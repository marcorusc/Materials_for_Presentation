## MaBoSS Functions for Loading Bnet Files

MaBoSS provides three functions to work with bnet files:

**API Reference:**
- `maboss.load()` - loads .bnd/.cfg file pairs or .bnet files
- `maboss.loadBNet()` - specifically loads .bnet files  
- `maboss.bnet_to_maboss()` - converts bnet content to MaBoSS simulation object
